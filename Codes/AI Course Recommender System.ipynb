### 8.11 Course Title Text Preprocessing

Before applying NLP techniques such as Bag of Words and TF-IDF, course titles need basic text preprocessing.

The preprocessing includes converting text to lowercase and removing unnecessary characters. This creates a cleaner and more consistent text representation for feature extraction.

```python
import re

def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df2_clean["processed_title"] = (
    df2_clean["TITLE"]
    .fillna("")
    .apply(preprocess_text)
)

display(
    df2_clean[[
        "TITLE",
        "processed_title"
    ]].head(10)
)
```

```python
# Compare original and processed title lengths

df2_clean["processed_title_length"] = (
    df2_clean["processed_title"].str.len()
)

display(
    df2_clean[[
        "TITLE",
        "processed_title",
        "processed_title_length"
    ]].head(10)
)
```

### 8.12 Stopword Analysis in Course Titles

Common words such as "the", "and", "of", and "in" can occur frequently in text without providing much information about the topic.

Stopword analysis helps identify these common terms before applying NLP techniques such as Bag of Words and TF-IDF. Removing such words can reduce unnecessary features and make the resulting text representation more focused on meaningful terms.

```python
from collections import Counter

# Common English stopwords
stop_words = {
    "a", "an", "the", "and", "or", "but", "if", "in", "on", "at",
    "to", "for", "of", "with", "by", "from", "is", "are", "was",
    "were", "be", "been", "this", "that", "these", "those", "as"
}

# Extract words from processed course titles
title_words = []

for title in df2_clean["processed_title"]:
    title_words.extend(title.split())

# Count all words
all_word_counts = Counter(title_words)

# Count stopwords
stopword_counts = Counter(
    word for word in title_words
    if word in stop_words
)

stopword_frequency = pd.DataFrame(
    stopword_counts.most_common(20),
    columns=["Stopword", "Frequency"]
)

display(stopword_frequency)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=stopword_frequency,
    x="Frequency",
    y="Stopword"
)

plt.title("Common Stopwords in Course Titles")
plt.xlabel("Frequency")
plt.ylabel("Stopword")

plt.tight_layout()

plt.savefig("course_title_stopwords.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

## 9.1 Bag of Words Feature Extraction

Bag of Words (BoW) is a text representation technique that converts textual data into numerical features.

In this project, BoW will be applied to the processed course titles to create a document-term matrix. Each row represents a course, while each column represents a word from the vocabulary. The values indicate how frequently each word occurs in a course title.

The resulting numerical representation will be used in later stages to calculate course similarity and generate content-based recommendations.

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create a lightweight Bag of Words model
bow_vectorizer = CountVectorizer(
    stop_words="english",
    max_features=1000
)

# Convert course titles into BoW features
bow_matrix = bow_vectorizer.fit_transform(
    df2_clean["processed_title"]
)

# Get vocabulary
bow_features = bow_vectorizer.get_feature_names_out()

print("Number of courses:", bow_matrix.shape[0])
print("Number of BoW features:", bow_matrix.shape[1])

print("\nFirst 15 vocabulary words:")
print(bow_features[:15])
```

```python
# Find the most frequent words

word_frequency = bow_matrix.sum(axis=0).A1

top_words = sorted(
    zip(bow_features, word_frequency),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("Top 10 words in course titles:")

for word, frequency in top_words:
    print(word, ":", frequency)
```

```python
# Visualize the top 10 BoW words

words = [x[0] for x in top_words]
frequencies = [x[1] for x in top_words]

plt.figure(figsize=(8, 4))
plt.barh(words[::-1], frequencies[::-1])

plt.title("Top 10 Words in Course Titles")
plt.xlabel("Frequency")
plt.ylabel("Word")

plt.tight_layout()
plt.savefig("top_10_bow_words.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 9.2 TF-IDF Feature Extraction

TF-IDF (Term Frequency–Inverse Document Frequency) converts course text into numerical features while giving greater importance to words that are useful for distinguishing one course from another.

Unlike Bag of Words, TF-IDF reduces the importance of words that appear frequently across many course titles.

A lightweight TF-IDF representation will be created using a maximum of 1,000 features. This representation will later be used for course similarity and recommendations.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=1000
)

tfidf_matrix = tfidf_vectorizer.fit_transform(
    df2_clean["processed_title"]
)

tfidf_features = tfidf_vectorizer.get_feature_names_out()

print("Courses:", tfidf_matrix.shape[0])
print("TF-IDF features:", tfidf_matrix.shape[1])
```

```python
# Find the most important words

tfidf_scores = tfidf_matrix.sum(axis=0).A1

top_tfidf = sorted(
    zip(tfidf_features, tfidf_scores),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("Top 10 TF-IDF words:")

for word, score in top_tfidf:
    print(word, ":", round(score, 3))
```

```python
# Display a small sample of the TF-IDF matrix

tfidf_sample = tfidf_matrix[:5, :10].toarray()

print("First 5 courses × first 10 TF-IDF features:")
print(tfidf_sample)
```

```python
tfidf_sample_df = pd.DataFrame(
    tfidf_matrix[:5, :10].toarray(),
    columns=tfidf_features[:10]
)

print("TF-IDF Matrix Sample:")
print(tfidf_sample_df)
```

```python
# Visualize top TF-IDF words

words = [x[0] for x in top_tfidf]
scores = [x[1] for x in top_tfidf]

plt.figure(figsize=(8, 4))
plt.barh(words[::-1], scores[::-1])

plt.title("Top 10 TF-IDF Words")
plt.xlabel("TF-IDF Score")
plt.ylabel("Word")

plt.tight_layout()
plt.savefig("top_10_tfidf_words.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 9.3 Course Similarity

Course similarity measures how closely two courses are related based on their textual content.

In this project, cosine similarity is used to compare the TF-IDF representation of courses. A higher cosine similarity score indicates that two courses have more similar content.

To keep the computation lightweight, similarity will be calculated only when recommendations are requested rather than creating a large similarity matrix for all courses.

```python
from sklearn.metrics.pairwise import cosine_similarity

def find_similar_courses(course_index, n=5):
    
    # Compare one course with all courses
    scores = cosine_similarity(
        tfidf_matrix[course_index],
        tfidf_matrix
    ).flatten()
    
    # Get the most similar courses
    similar_indices = scores.argsort()[::-1]
    
    # Remove the selected course
    similar_indices = similar_indices[
        similar_indices != course_index
    ][:n]
    
    return similar_indices, scores[similar_indices]
```

```python
# Test similarity for the first course

course_index = 0

similar_indices, similarity_scores = find_similar_courses(
    course_index,
    n=5
)

print("Selected course:")
print(df2_clean.iloc[course_index]["TITLE"])

print("\nSimilar courses:")

for index, score in zip(similar_indices, similarity_scores):
    print(
        df2_clean.iloc[index]["TITLE"],
        "->",
        round(score, 3)
    )
```

## 9.4 Course Similarity Recommendations

For a given course, the system identifies the courses with the highest cosine similarity scores and presents them as recommendations. The selected course itself is excluded from the recommendation list.

This provides a simple content-based recommendation approach based on course title information.

```python
# Recommend similar courses

selected_index = 0

similar_indices, similarity_scores = find_similar_courses(
    selected_index,
    n=5
)

print("Selected Course:")
print(df2_clean.iloc[selected_index]["TITLE"])

print("\nRecommended Courses:")

for index, score in zip(similar_indices, similarity_scores):
    print(
        f"{df2_clean.iloc[index]['TITLE']} "
        f"(Similarity: {score:.3f})"
    )
```

## 9.6 User Profile and Course Similarity

The user profiles created from course genres can be compared with the genre characteristics of courses.

For each learner, the system identifies courses whose genre composition is closest to the learner's interests. Cosine similarity is used to measure the similarity between the learner profile and each course profile.

Only a single selected user is evaluated at a time to keep the computation lightweight.

```python
from sklearn.metrics.pairwise import cosine_similarity

# Create a course genre matrix
course_profiles = (
    df1_clean
    .set_index("COURSE_ID")[genre_columns]
)

# Select one user
selected_user = user_profiles.index[0]

# Get the user's profile
user_vector = user_profiles.loc[
    selected_user
].values.reshape(1, -1)

# Calculate similarity with all courses
scores = cosine_similarity(
    user_vector,
    course_profiles.values
).flatten()

# Get top 5 courses
top_indices = scores.argsort()[::-1][:5]

print("Selected User:", selected_user)
print("\nRecommended courses based on user interests:")

for index in top_indices:
    course_id = course_profiles.index[index]
    
    title = df2_clean.loc[
        df2_clean["COURSE_ID"] == course_id,
        "TITLE"
    ].iloc[0]
    
    print(f"{title} (Similarity: {scores[index]:.3f})")
```

## 9.7 User Profile Clustering

User profile clustering groups learners with similar course interests.

K-Means clustering is used to identify a small number of learner groups based on their genre preferences. This can help the recommendation system understand different types of learners without performing expensive calculations on the complete interaction dataset.

A small number of clusters will be tested, and the silhouette score will be used to select the better clustering configuration.

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Use the user profile genre features
X_users = user_profiles.values

# Test a small number of clusters
results = []

for k in range(2, 5):
    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )
    
    labels = model.fit_predict(X_users)
    
    score = silhouette_score(
        X_users,
        labels
    )
    
    results.append((k, score))

print("Clustering results:")

for k, score in results:
    print(f"Clusters: {k} | Silhouette Score: {score:.3f}")
```

```python
# Select the best cluster count

best_k = max(
    results,
    key=lambda x: x[1]
)[0]

user_cluster_model = KMeans(
    n_clusters=best_k,
    random_state=RANDOM_STATE,
    n_init=10
)

user_clusters = user_cluster_model.fit_predict(
    X_users
)

print("Selected number of clusters:", best_k)
```

```python
# Small cluster-size graph

cluster_counts = pd.Series(
    user_clusters
).value_counts().sort_index()

plt.figure(figsize=(6, 4))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.title("Learner Distribution Across Clusters")
plt.xlabel("Cluster")
plt.ylabel("Number of Users")

plt.tight_layout()

plt.savefig(
    "user_profile_clusters.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## 10.1 User-Course Rating Matrix

A user-course rating matrix represents the interactions between learners and courses.

Each row represents a learner, each column represents a course, and the values represent the ratings given by learners.

This matrix provides the basic structure required for collaborative filtering. To keep the computation lightweight, only a small sample of users and courses will be used for visualization, while the complete interaction data remains available for later modeling.

```python
# Create the user-course rating matrix

user_course_matrix = df3_clean.pivot_table(
    index="user",
    columns="item",
    values="rating",
    aggfunc="mean",
    fill_value=0
)

print("Users:", user_course_matrix.shape[0])
print("Courses:", user_course_matrix.shape[1])
```

```python
# Display only a small sample

print("Sample of the user-course rating matrix:")
print(user_course_matrix.iloc[:5, :10])
```

```python
# Calculate matrix density

total_cells = (
    user_course_matrix.shape[0] *
    user_course_matrix.shape[1]
)

observed_cells = (
    user_course_matrix != 0
).sum().sum()

density = (
    observed_cells / total_cells
) * 100

print("Matrix density:", round(density, 4), "%")
```

## 10.2 KNN Collaborative Filtering

K-Nearest Neighbors (KNN) recommends courses by finding learners with similar rating patterns.

In this approach, the rating vectors of learners are compared using cosine similarity. Courses that similar learners have rated highly can then be considered as potential recommendations.

To keep the computation lightweight, KNN is applied to a small sample of users rather than performing expensive calculations across the complete user population.

```python
from sklearn.neighbors import NearestNeighbors

# Use a small sample of users
sample_users = user_course_matrix.iloc[:300]

# KNN model using cosine distance
knn_model = NearestNeighbors(
    n_neighbors=6,
    metric="cosine",
    algorithm="brute"
)

knn_model.fit(sample_users)

# Select one user
selected_user_index = 0

distances, indices = knn_model.kneighbors(
    sample_users.iloc[
        selected_user_index:selected_user_index + 1
    ]
)

print("Selected User:")
print(sample_users.index[selected_user_index])

print("\nSimilar Users:")

for i, distance in zip(indices[0][1:], distances[0][1:]):
    print(
        sample_users.index[i],
        "-> Similarity:",
        round(1 - distance, 3)
    )
```

```python
# Get similar users
similar_user_indices = indices[0][1:]

# Ratings of similar users
similar_user_ratings = sample_users.iloc[
    similar_user_indices
]

# Average their course ratings
course_scores = similar_user_ratings.mean()

# Exclude courses already rated by the selected user
selected_user_ratings = sample_users.iloc[
    selected_user_index
]

unrated_courses = selected_user_ratings[
    selected_user_ratings == 0
].index

recommendations = course_scores[
    unrated_courses
].sort_values(
    ascending=False
).head(5)

print("Recommended Courses:")

for course_id, score in recommendations.items():
    title = df2_clean.loc[
        df2_clean["COURSE_ID"] == course_id,
        "TITLE"
    ]

    if len(title) > 0:
        print(
            f"{title.iloc[0]} "
            f"(Score: {score:.2f})"
        )
```

```python
# Average rating of recommended courses

if len(recommendations) > 0:
    print(
        "Average recommendation score:",
        round(recommendations.mean(), 2)
    )
```

## 10.3 NMF Collaborative Filtering

Non-Negative Matrix Factorization (NMF) decomposes the user-course rating matrix into smaller non-negative matrices.

The resulting latent factors represent hidden patterns in learner preferences and course characteristics. These factors can be used to reconstruct missing ratings and generate recommendations.

NMF is suitable for this project because course ratings are non-negative. To keep the computation lightweight, a limited number of components will be used.

```python
from sklearn.decomposition import NMF

# Use a small sample of the rating matrix
nmf_data = user_course_matrix.iloc[:300, :300]

# Create NMF model
nmf_model = NMF(
    n_components=5,
    init="random",
    random_state=RANDOM_STATE,
    max_iter=100
)

# Learn latent factors
user_factors = nmf_model.fit_transform(nmf_data)
course_factors = nmf_model.components_

print("User factor shape:", user_factors.shape)
print("Course factor shape:", course_factors.shape)
print("Reconstruction error:", round(nmf_model.reconstruction_err_, 3))
```

```python
# Reconstruct the rating matrix
predicted_ratings = np.dot(
    user_factors,
    course_factors
)

# Recommendations for the first user
user_ratings = nmf_data.iloc[0].values

unrated = user_ratings == 0

scores = predicted_ratings[0].copy()
scores[~unrated] = -1

top_courses = np.argsort(scores)[::-1][:5]

print("Top NMF recommended course IDs:")

for idx in top_courses:
    print(
        nmf_data.columns[idx],
        "-> Predicted score:",
        round(scores[idx], 2)
    )
```

## 10.4 Neural Network Collaborative Filtering

Neural networks can learn nonlinear relationships between learners and courses.

In this approach, user IDs and course IDs are represented using embedding layers. These embeddings are combined and passed through a small neural network to predict the learner's rating for a course.

A lightweight architecture is used to keep training fast while demonstrating a deep-learning-based collaborative filtering approach.

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# Use a small sample for lightweight training
nn_data = df3_clean.head(5000).copy()

# Convert user and course IDs to integer indices
user_ids = nn_data["user"].astype("category")
course_ids = nn_data["item"].astype("category")

user_index = user_ids.cat.codes.values
course_index = course_ids.cat.codes.values

n_users = user_ids.cat.categories.size
n_courses = course_ids.cat.categories.size

# User and course inputs
user_input = layers.Input(shape=(1,))
course_input = layers.Input(shape=(1,))

# Embeddings
user_embedding = layers.Embedding(
    n_users, 8
)(user_input)

course_embedding = layers.Embedding(
    n_courses, 8
)(course_input)

# Flatten embeddings
user_vector = layers.Flatten()(user_embedding)
course_vector = layers.Flatten()(course_embedding)

# Combine
combined = layers.Concatenate()([
    user_vector,
    course_vector
])

# Small neural network
hidden = layers.Dense(16, activation="relu")(combined)
output = layers.Dense(1)(hidden)

model = Model(
    inputs=[user_input, course_input],
    outputs=output
)

model.compile(
    optimizer="adam",
    loss="mse"
)

model.fit(
    [user_index, course_index],
    nn_data["rating"].values,
    epochs=3,
    batch_size=64,
    verbose=1
)

print("Neural network collaborative filtering model trained.")
```

```python
# Evaluate training error

loss = model.evaluate(
    [user_index, course_index],
    nn_data["rating"].values,
    verbose=0
)

print("Neural Network MSE:", round(loss, 4))
```

```python
history = model.fit(
    [user_index, course_index],
    nn_data["rating"].values,
    epochs=3,
    batch_size=64,
    verbose=1
)
```

```python
# Neural network training loss

plt.figure(figsize=(7, 4))

plt.plot(
    history.history["loss"],
    marker="o"
)

plt.title("Neural Network Training Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

plt.tight_layout()

plt.savefig(
    "neural_network_training_loss.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## 11.3 F1-Score

F1-score combines precision and recall into a single metric.

It is useful when both recommending relevant courses and finding relevant courses are important.

```python
from sklearn.metrics import f1_score

f1 = f1_score(
    actual,
    predicted
)

print("F1-Score:", round(f1, 3))
```

## 11.4 Recommendation Diversity

Recommendation diversity measures how varied the recommended courses are.

A recommendation list containing courses from different genres can provide learners with broader learning options, while a list containing very similar courses may be less diverse.

In this project, diversity is measured using the number of different course genres represented in a small recommendation list.

```python
# Calculate diversity of the KNN recommendations

recommended_course_ids = recommendations.index

recommended_genres = df1_clean[
    df1_clean["COURSE_ID"].isin(recommended_course_ids)
][genre_columns]

# Count different genres represented
genre_presence = (recommended_genres.sum(axis=0) > 0).sum()

print(
    "Different genres in recommendations:",
    genre_presence
)

print(
    "Recommendation diversity:",
    round(
        genre_presence / len(genre_columns),
        3
    )
)
```

```python
# Visualize recommendation diversity

plt.figure(figsize=(6, 4))

plt.bar(
    ["Recommendation Diversity"],
    [genre_presence]
)

plt.title("Recommendation Diversity")
plt.ylabel("Number of Genres")

plt.tight_layout()

plt.savefig(
    "recommendation_diversity.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## 11.5 Catalog Coverage

Catalog coverage measures how much of the available course catalog is represented in the recommendations.

Higher coverage means that the recommendation system is able to recommend a wider range of courses instead of repeatedly recommending only a small group of popular courses.

In this project, catalog coverage is calculated from the courses recommended by the KNN model.

```python
# Calculate catalog coverage

total_courses = df2_clean["COURSE_ID"].nunique()

recommended_courses = len(
    set(recommendations.index)
)

catalog_coverage = (
    recommended_courses / total_courses
) * 100

print("Total courses:", total_courses)
print("Recommended courses:", recommended_courses)
print(
    "Catalog coverage:",
    round(catalog_coverage, 3),
    "%"
)
```

```python
# Visualize catalog coverage

plt.figure(figsize=(6, 4))

plt.bar(
    ["Recommended", "Not Recommended"],
    [
        recommended_courses,
        total_courses - recommended_courses
    ]
)

plt.title("Catalog Coverage")
plt.ylabel("Number of Courses")

plt.tight_layout()

plt.savefig(
    "catalog_coverage.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
